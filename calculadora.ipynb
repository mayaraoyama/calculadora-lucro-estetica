```python
continuar = "sim"
while continuar == "sim":
    Procedimento = input("Qual procedimento? ")
    print(Procedimento)
    quantidade_sessao = int(input("Quantidade realizada "))
    print(quantidade_sessao)
    valor_sessao = float(input("Valor da sessão "))
    print (valor_sessao)
    valor_produto = float(input("Valor do produto "))
    print (valor_produto)
    bruto = float (valor_sessao-valor_produto)
    print (bruto)
    total_bruto = float (bruto*quantidade_sessao)
    print (total_bruto)
    desconto_avista = float (valor_sessao-(valor_sessao*0.1))
    print (desconto_avista)
    lucro = float (desconto_avista-valor_produto)
    print (lucro)
    if lucro >300:
        print ("Lucro satisfatório")
    else:
        print ("Revisar precificação")
    continuar = input("Deseja calcular outro procedimento? (sim/não) ")


```
